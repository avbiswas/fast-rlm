# Graph-native FastRLM: WebQSP

This proof of concept merges the first 50 WebQSP test graphs in the notebook, passes the resulting `nx.MultiDiGraph` through FastRLM's graph channel, and performs subgraph extraction and answer detection inside the RLM runtimes.

Install the example dependencies with `pip install -e '.[graph]' datasets pydantic`.

```python
import os

import fast_rlm
import networkx as nx
from datasets import load_dataset
from fast_rlm import RLMConfig
from pydantic import BaseModel

DATASET_NAME = "rmanluo/RoG-webqsp"
MODEL = os.getenv("RLM_MODEL", "minimax-m3:cloud")
os.environ.setdefault("RLM_MODEL_BASE_URL", "http://localhost:11434/v1")
os.environ.setdefault("RLM_MODEL_API_KEY", "ollama")
```

```python
test_rows = load_dataset(DATASET_NAME, split="test").select(range(50))
example = test_rows[0]

combined_graph = nx.MultiDiGraph()
seen_triples = set()
for row in test_rows:
    for subject, relation, object_ in row["graph"]:
        triple = (str(subject), str(relation), str(object_))
        if triple in seen_triples:
            continue
        seen_triples.add(triple)
        combined_graph.add_edge(
            triple[0],
            triple[2],
            key=triple[1],
            relation=triple[1],
        )
```

```python
class WebQSPAnswer(BaseModel):
    answer_entities: list[str]


config = RLMConfig(
    primary_agent=MODEL,
    sub_agent=MODEL,
    max_depth=1,
    max_calls_per_subagent=8,
    max_global_calls=12,
)

instruction = """
You are the main graph RLM. A complete NetworkX MultiDiGraph is available as `graph`.

You MUST make exactly one graph child call using:

    subgraph = await graph_query(
        {"question": context["question"], "q_entity": context["q_entity"]},
        graph=graph,
        instruction=CHILD_INSTRUCTION,
    )

CHILD_INSTRUCTION must tell the child to do all of the following inside its RLM REPL:
- Confirm `graph` is a real NetworkX MultiDiGraph.
- Start from the supplied q_entity labels.
- Traverse with NetworkX functions, including out_edges(..., keys=True, data=True),
  in_edges(..., keys=True, data=True), neighbor or degree inspection as useful, and
  edge_subgraph(...).copy().
- Use the question to select only relations and neighboring entities relevant to answering it.
- Return a relevant NetworkX subgraph with at most 20 edges by calling GRAPH_FINAL(subgraph).
- Never answer the question and never call ordinary FINAL.

After graph_query returns, confirm `subgraph` is a real NetworkX graph. Inspect its nodes and
edges with NetworkX and answer only from facts present in it. Return exact graph node labels in
the required answer_entities schema using ordinary FINAL. Do not answer from outside knowledge.
""".strip()
```

```python
result = fast_rlm.run(
    {
        "question": str(example["question"]),
        "q_entity": [str(entity) for entity in example["q_entity"]],
    },
    graph=combined_graph,
    config=config,
    output_schema=WebQSPAnswer,
    instruction=instruction,
    verbosity="silent",
)

print("Prediction:", result["results"]["answer_entities"])
print("Golden answer:", [str(answer) for answer in example["answer"]])
```
